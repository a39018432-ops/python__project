## Objective
Analyze internship and job data using Python, NumPy and Pandas.

```python
import pandas as pd
import numpy as np
```

```python
data={
'Company':['TCS','Infosys','Wipro','Google','Microsoft','Amazon','IBM','Accenture'],
'Role':['Intern','Intern','Job','Intern','Job','Job','Intern','Job'],
'Stipend_Salary':[15000,18000,450000,40000,1200000,800000,20000,600000],
'Skills':['Python','Python','SQL','Python','Python','Java','Python','SQL']
}
df=pd.DataFrame(data)
df
```

```python
print('Average Package/Stipend:',np.mean(df['Stipend_Salary']))
print('Maximum:',np.max(df['Stipend_Salary']))
print('Minimum:',np.min(df['Stipend_Salary']))
```

```python
print(df.groupby('Role')['Stipend_Salary'].mean())
```

```python
python_jobs=df[df['Skills']=='Python']
python_jobs
```

```python
df['Level']=np.where(df['Stipend_Salary']>500000,'High','Normal')
df
```
